```python
from pandas import *
from scipy import stats
import numpy as np
```

```python
df = read_csv('/content/customer_demographics.csv')
```

```python
df.head()
```

```python
clean_df = df.dropna()
```

```python
clean_df.head()
```

```python
# convert age range to numeric value
def convert_age_range_to_numeric(age_range):
    if age_range == '70+':
        return 70
    else:
        mini, maxi = age_range.split('-')
        return (int(mini) + int(maxi)) / 2

# Apply the conversion function to create a new column
clean_df['age_numeric'] = clean_df['age_range'].apply(convert_age_range_to_numeric)

print(clean_df)
```

```python
# Shapiro-Wilk Test for Normalityif age_range_normality.pvalue > 0.05:
age_range_normality = stats.shapiro(clean_df['age_numeric'])
print(age_range_normality)
if age_range_normality.pvalue > 0.05:
    print("The data is normally distributed :fail to reject hypothesis)")
else:
    print("The data is not normally distributed : reject hypothesis)")

```

```python
#Pearson's chi-square test of goodness-of-fit
age_range_normality= stats.normaltest(clean_df['age_numeric'])
print(age_range_normality)
```

```python
#find the skewness of the sample
stats.skew(clean_df['age_numeric'], bias=True)
```

```python
#kurtosis of the sample
stats.kurtosis(clean_df['age_numeric'], fisher=True)
```

```python
#standard deviation
std_deviation = clean_df['age_numeric'].std()
print(std_deviation)
```

```python
#convert family size data to numeric values
def convert_size_to_numeric(family_size):
    if family_size == '5+':
        return 5
    else:
        return int(family_size)

# Apply the conversion function to create a new column
clean_df['family_size_numeric'] = clean_df['family_size'].apply(convert_size_to_numeric)

print(clean_df)
```

```python
family_size_normality = stats.shapiro(clean_df['family_size_numeric'])
print(family_size_normality)
if family_size_normality.pvalue > 0.05:
    print("The data is normally distributed :fail to reject hypothesis)")
else:
    print("The data is not normally distributed : reject hypothesis)")
```

```python
#find the skewness of the sample
stats.skew(clean_df['family_size_numeric'], bias=True)
```

```python
#kurtosis of the sample
stats.kurtosis(clean_df['family_size_numeric'], fisher=True)
```

```python
#standard deviation
std_deviation = clean_df['family_size_numeric'].std()
print(std_deviation)
```
